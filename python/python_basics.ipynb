## What are Comprehensions?

Comprehensions are a Pythonic way to create collections (lists, dictionaries, sets) in a single, readable line. They combine loops and conditional logic into a concise syntax that's both efficient and expressive.

**Basic Syntax Pattern:**

```
[expression for item in iterable if condition]
```



## 1. List Comprehensions

### Definition

Creates a new list by applying an expression to each item in an iterable, optionally filtering with conditions.

### Syntax

```python
[expression for item in iterable if condition]
```

### Example: Square Numbers

**Conventional Way:**

````python
# Create a list of squares for numbers 1-5
squares = []
for i in range(1, 6):
    squares.append(i ** 2)
print(squares)  # [1, 4, 9, 16, 25]
````

````python
squares = [i ** 2 for i in range(1, 6)]
print(squares)
````

````python
# WAP which returns a list of lengths of the words in a list.
# ["Rohan Kumar", "Mohan Sharma", "Steve", "John Smith"]
#

names = ["Rohan Kumar", "Mohan Sharma", "Steve", "John Smith"]

def get_lengths(lst:List[str]):
    res = []
    for name in lst:
        res.append(len(name))

    return res

get_lengths(names)
````

List comprehensions are generally faster than for loops in Python because they are optimized at the C-level within the interpreter.


The performance gain is primarily due to:

- **Avoidance of Method Lookups:** In a standard `for` loop, you typically call `list.append()` in every iteration. This requires looking up the `append` attribute on the list object and then calling it as a function, which adds overhead.
- **Specialized Bytecode:** List comprehensions use a dedicated bytecode instruction called `LIST_APPEND`. This instruction is a fast, internal operation that appends the current value directly to the result list without the need for a full function call or attribute lookup.
- **Internal Loop Optimization:** The iteration in a list comprehension is handled by the Python interpreter's internal C code, which is more efficient than the high-level Python loop structure.
- **Reduced Instruction Count:** List comprehensions typically involve fewer bytecode instructions overall compared to an equivalent `for` loop, leading to faster execution times.

Subset

A set `A` is a **subset** of another set `B` if all elements of set `A` are also present in set `B`.
- **Example**: If `A = {1, 2}` and `B = {1, 2, 3, 4}`, `A` is a subset of `B`.
- **Note**: A set is a subset of itself, and the empty set is a subset of every set.

Python Implementation

You can check for a subset relationship using:

- **The `.issubset()` method**: `A.issubset(B)` returns `True` if `A` is a subset of `B`.
- **The `<=` operator**: `A <= B` returns `True` if `A` is a subset of, or equal to, `B`.
- **The `<` operator**: `A < B` checks for a _proper_ subset (all elements of `A` are in `B`, but `A` is not equal to `B`).

```python
A = {1, 2}
B = {1, 2, 3, 4}

print(A.issubset(B))  # Output: True
print(A <= B)         # Output: True
print(A < B)          # Output: True (proper subset)

C = {1, 2, 3, 4}
print(C < B)          # Output: False (not a proper subset, as they are equal)

```


Superset

A set `A` is a **superset** of another set `B` if `A` contains all elements of set `B` (and possibly more). This is the inverse relationship of a subset.

- **Example**: Using the previous sets, `B = {1, 2, 3, 4}` is a superset of `A = {1, 2}`.
- **Note**: A set is a superset of itself.


Python Implementation

You can check for a superset relationship using:

- **The `.issuperset()` method**: `A.issuperset(B)` returns `True` if `A` is a superset of `B`.
- **The `>=` operator**: `A >= B` returns `True` if `A` is a superset of, or equal to, `B`.
- **The `>` operator**: `A > B` checks for a _proper_ superset (all elements of `B` are in `A`, but `A` is not equal to `B`).

```python
A = {1, 2}
B = {1, 2, 3, 4}

print(B.issuperset(A)) # Output: True
print(B >= A)          # Output: True
print(B > A)           # Output: True (proper superset)

C = {1, 2, 3, 4}
print(C > B)           # Output: False (not a proper superset, as they are equal)

```

String formatting in Python is a powerful way to create clean and readable output, especially when working with numbers, dates, and other data types. Python provides several methods for string formatting, including the older `str.format()` method, the `f-string` (formatted string literals), and the `%` formatting (which is now less commonly used but still supported).

This guide will focus primarily on the most modern and powerful method: **f-string formatting** (introduced in Python 3.6). We'll explore the various ways to format numbers, control alignment, padding, and more, using both f-strings and `str.format()`.


### 1. **Basic String Formatting with f-strings**

An f-string allows you to embed expressions inside string literals, using curly braces `{}`. It automatically evaluates the expression inside the braces and inserts the result into the string.

**Syntax:**

```python
f"some string {expression}"
```

Example:

```python
name = "Alice"
age = 30
print(f"Hello, my name is {name} and I am {age} years old.")
```

Output:

```
Hello, my name is Alice and I am 30 years old.
```


### 2. **Number Formatting in f-strings**

When dealing with numbers, f-strings allow precise control over how they are formatted. You can specify the number of decimal places, padding, alignment, and more. This is controlled by the **format specifiers**.

### 2.1 **Integer Formatting:**

Use the `d` format specifier to format integers.

Example:

````python
number = 1234

print(f"Integer: {number:d}")
````

Use the `f` format specifier for floating-point numbers. You can control the number of decimal places.

Example:

````python
pi = 3.141592653589793
print(f"Pi rounded to 2 decimal places:{pi:.2f}")
````

### 2.3 **Scientific Notation:**

Use the `e` format specifier for scientific notation.

Example:

````python
large_number = 12668787
print(f"Scientific notation:{large_number:.2e}")
````

To format a number as a percentage, use the `%` format specifier. This multiplies the value by 100 and appends a `%` symbol.

Where:

- **`datetime_object`**: A `datetime` object (e.g., `datetime.now()` or any `datetime` object you create).

````python
text = "Hello World! How Are You?"

print("Original:", text)
print()

# lower() - Convert to lowercase
print("lower():", text.lower())

# upper() - Convert to uppercase
print("upper():", text.upper())

# capitalize() - First char upper, rest lower
print("capitalize():", text.capitalize())

# title() - Title Case Each Word
print("title():", text.title())

# swapcase() - Swap case of each character
print("swapcase():", text.swapcase())


# casefold() - Aggressive lowercase for caseless matching
german = "Straße"  # German sharp S
print(f"\ncasefold() vs lower(): '{german}' -> '{german.casefold()}' vs '{german.lower()}'")



````

````python
text = "Python is awesome. Python is powerful."

print(f"Text: '{text}'\n")
# find() - Returns index of first occurrence, -1 if not found
print(f"find('Python'): {text.find('Python')}")

print(f"find('Python', 10): {text.find('Python', 10)}")
print(f"find('Java'): {text.find('Java')}")

# rfind() - Find from right
print(f"\nrfind('Python'): {text.rfind('Python')}")


# index() - Like find() but raises ValueError if not found
print(f"\nindex('awesome'): {text.index('awesome')}")

try:
    text.index('Java')
except ValueError as e:
    print(f"index('Java') raises: {e}")

# count() - Count non-overlapping occurrences

print(f"\ncount('o'): {text.count('o')}")
print(f"count('Python'): {text.count('Python')}")



# startswith() and endswith()
print(f"\nstartswith('Python'): {text.startswith('Python')}")

print(f"endswith('powerful.'): {text.endswith('powerful.')}")
print(f"startswith(('Python', 'Java')): {text.startswith(('Python', 'Java'))}")

````

## **Validation Methods (*is* methods)**

````python
# All is* methods return Boolean
print("isalpha() - All alphabetic:")

print(f"  'Hello'.isalpha() = {'Hello'.isalpha()}")
print(f"  'Hello123'.isalpha() = {'Hello123'.isalpha()}")

print("\nisdigit() - All digits:")
print(f"  '12345'.isdigit() = {'12345'.isdigit()}")
print(f"  '123.45'.isdigit() = {'123.45'.isdigit()}")


print("\nisalnum() - All alphanumeric:")
print(f"  'Hello123'.isalnum() = {'Hello123'.isalnum()}")
print(f"  'Hello 123'.isalnum() = {'Hello 123'.isalnum()}")

print("\nisspace() - All whitespace:")
print(f"  '   '.isspace() = {'   '.isspace()}")
print(f"  ' a '.isspace() = {' a '.isspace()}")


print("\nislower() / isupper():")
print(f"  'hello'.islower() = {'hello'.islower()}")
print(f"  'HELLO'.isupper() = {'HELLO'.isupper()}")
print(f"  'Hello'.islower() = {'Hello'.islower()}")


print("\nistitle() - Title Case:")
print(f"  'Hello World'.istitle() = {'Hello World'.istitle()}")
print(f"  'Hello world'.istitle() = {'Hello world'.istitle()}")

print("\nisdecimal() / isnumeric() / isdigit():")
test_str = "½"
print(f"  String: '{test_str}'")
print(f"  isdecimal() = {test_str.isdecimal()}")
print(f"  isdigit() = {test_str.isdigit()}")
print(f"  isnumeric() = {test_str.isnumeric()}")


print("\nisidentifier() - Valid Python identifier:")
print(f"  'variable_name'.isidentifier() = {'variable_name'.isidentifier()}")
print(f"  '2variable'.isidentifier() = {'2variable'.isidentifier()}")
print(f"  'for'.isidentifier() = {'for'.isidentifier()}")


````

## Modification Methods

````python
# Remember: All return NEW strings, original unchanged

# replace() - Replace substring
text = "Hello World World"
print(f"Original: '{text}'")
print(f"replace('World', 'Python'): '{text.replace('World', 'Python')}'")
print(f"replace('World', 'Python', 1): '{text.replace('World', 'Python', 1)}'")

# strip() / lstrip() / rstrip() - Remove whitespace
text = "   Hello World   "
print(f"\nOriginal: '{text}'")
print(f"strip(): '{text.strip()}'")
print(f"lstrip(): '{text.lstrip()}'")
print(f"rstrip(): '{text.rstrip()}'")

# Can strip specific characters
text = "xxxHelloxxx"
print(f"\n'{text}'.strip('x') = '{text.strip('x')}'")

# removeprefix() / removesuffix() (Python 3.9+)
filename = "document.txt"
print(f"\n'{filename}'.removesuffix('.txt') = '{filename.removesuffix('.txt')}'")

url = "https://example.com"
print(f"'{url}'.removeprefix('https://') = '{url.removeprefix('https://')}'")
````
