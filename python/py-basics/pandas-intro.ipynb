### What is Pandas?

**Pandas** (Panel Data / Python Data Analysis Library) is an open-source library built on top of NumPy that provides **high-performance, easy-to-use data structures and data analysis tools** for Python.

Created by Wes McKinney in 2008 while at AQR Capital Management, Pandas was born out of the need to have a flexible, expressive tool for financial data analysis — something between a spreadsheet (Excel) and a database (SQL), but with the full power of Python.

At its core, Pandas gives you:
- **DataFrame**: a 2D labeled table (like a spreadsheet or SQL table)
- **Series**: a 1D labeled array (like a single column)
- Rich tools for reading/writing dozens of file formats
- Powerful indexing, merging, grouping, reshaping, and time-series operations

A Series is a one-dimensional array-like object, while a DataFrame is a two-dimensional, size-mutable, and potentially heterogeneous tabular data structure with labeled axes (rows and columns).

### Series

A Series is a one-dimensional labeled array capable of holding any data type. It is similar to a column in a spreadsheet or a single variable in statistics.

```python
data = [12,24,35,48,59]
s1 = pd.Series(data)
s1
```

```python
type(s1)
```

```python
# Creating a Series from a list
# think of series as a list with index

ages = pd.Series([25, 30, 25, 40],
                  index= ['Rajesh', 'Sai', 'Nik', 'vijay'])
ages
```

#### Labels
If nothing else is specified, the values are labeled with their index number. First value has index 0, second value has index 1 etc.

This label can be used to access a specified value.

```python
ages["Sai"]
```

```python
print(ages.mean())
```

### DataFrame

A DataFrame is a 2-dimensional labeled data structure with columns that can be of different types. It's similar to a spreadsheet, SQL table, or a dictionary of Series objects.

#### Creating a DataFrame from a dictionary:

```python
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'Age': [24, 27, 22, 32, 29],
    'City': ['New York', 'Boston', 'Chicago', 'Seattle', 'San Francisco'],
    'Salary': [65000, 72000, 59000, 82000, 75000]
}

df3 = pd.DataFrame(data)

df3
```
